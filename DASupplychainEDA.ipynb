```python
warehouse_metrics = (df.groupby('Warehouse_ID').agg({'Shipping_Time_Days': 'mean','Order_Processing_Time': 'mean','Employee_Count': 'mean'}))
```

```python
final_warehouses = warehouse_metrics[(warehouse_metrics['Shipping_Time_Days'] < warehouse_metrics['Order_Processing_Time'])
 & (warehouse_metrics['Time_Difference'] > 2)]
```

```python
final_warehouses.sort_values('Time_Difference', ascending=False)
```
